```python
# ==========================================
# Python 猜拳遊戲：請在「你的程式碼」下方完成填空
# ==========================================

# 1. 冒險者登入
# ✏️ 你的程式碼：請使用 input() 詢問「請輸入你的冒險者名字：」，並存放到變數 name 中
name = input("請輸入你的冒險者名字：")

# ✏️ 你的程式碼：請使用 input() 詢問「請選擇你的使用裝備（例如：黃金鍵盤、魔法魔杖）：」，並存放到變數 equipment 中
equipment = input("請選擇你的使用裝備（例如：黃金鍵盤、魔法魔杖）：")

# 印出前置劇情 (此段不需修改)
print("\n=== 系統廣播 ===")
print(f"歡迎勇敢的冒險者【{name}】！")
print(f"你已經裝備了【{equipment}】，準備好迎接第一關的 【猜拳】 挑戰！")


import random # 召喚處理「隨機」的魔法模組

print("\n=== ✊✋✌️ 剪刀石頭布 終極大對決 ===")
print("規則：先獲得 3 勝的一方就是最終贏家！\n")

# 戰鬥前置作業：設定變數來記錄雙方的勝場數與回合數
player_wins = 0
computer_wins = 0
round_num = 1

# 設定電腦可以出的拳
options = ["剪刀", "石頭", "布"]

# 2. 迴圈條件設定
# ✏️️ 你的程式碼：只要玩家和電腦的勝場「都小於 3」，遊戲就繼續 (提示：使用 < 搭配 and)
while player_wins < 3 and computer_wins < 3:
    print(f"--- 第 {round_num} 回合 ---")
    print(f"🏆 目前戰況 - 你：{player_wins} 勝 | 電腦：{computer_wins} 勝")

    # 3. 玩家出拳
    # ✏️ 你的程式碼：使用 input() 詢問「請出拳 (輸入 剪刀、石頭 或 布)：」，存入 player_choice
    player_choice = input("請出拳 (輸入 剪刀、石頭 或 布)：")

    # 檢查玩家是否輸入錯誤 (此段不需修改)
    if player_choice != "剪刀" and player_choice != "石頭" and player_choice != "布":
        print("⚠️ 哎呀！指令錯誤，請輸入 剪刀、石頭 或 布。\n")
        continue # 直接重新開始新的一回合

    # 4. 電腦隨機出拳
    # ✏️ 你的程式碼：使用 random.choice() 從 options 串列中隨機挑選一個，存入 computer_choice
    computer_choice = random.choice(options)

    print(f"🤖 電腦出了：{computer_choice}")

    # 5. 勝負判斷
    # ✏️ 你的程式碼：如果 玩家出拳「等於」電腦出拳，代表平手 (提示：使用 ==)
    if player_choice == computer_choice:
        print("平手！沒有人得分。\n")
    elif (player_choice == "剪刀" and computer_choice == "布") or \
         (player_choice == "石頭" and computer_choice == "剪刀") or \
         (player_choice == "布" and computer_choice == "石頭"):
        print("🎉 這回合你贏了！\n")
        # ✏️️ 你的程式碼：請將 玩家勝場數 (player_wins) 加 1
        player_wins = player_wins + 1
    else:
        print("💥 這回合電腦贏了！\n")
        # ✏️ 你的程式碼：請將 電腦勝場數 (computer_wins) 加 1
        computer_wins = computer_wins + 1

    round_num = round_num + 1


# 6. 最終結算
print("=== 🌟 最終結果 🌟 ===")
# ✏️ 你的程式碼：如果 玩家勝場數「等於」3，印出玩家勝利的訊息
if player_wins == 3:
    print(f"🏆 恭喜你以 {player_wins} : {computer_wins} 擊敗了電腦，成為剪刀石頭布大師！")
else:
    print(f"💻 很可惜，電腦以 {computer_wins} : {player_wins} 獲得了最終勝利，下次再來挑戰吧！")

# ==========================================
```
